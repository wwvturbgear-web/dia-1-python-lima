```python
import pandas as pd
datos = {
    'Nombre':['Willian', 'Ana', 'Carlos'],
    'Edad' : [26, 24, 31],
    'Ciudad' : ['Lima', 'Cusco', 'Arequipa']
}
df = pd.DataFrame(datos)
print(df)
df. describe()
```

```python
import matplotlib.pyplot as plt
df_Lima = df[df['Ciudad'] == 'Lima']
print("Gente de Lima:")
print(df_Lima)

plt.bar(df['Nombre'], df['Edad'], color='#3776AB')
plt.title('Edades del equipo de Willian')
plt.ylabel('Edad')
plt.show()
```
